# Import of title.principals.tsv.gz file

```python
import pandas as pd
# we make pandas decompressed the file title.principals.tsv.gz 
# for reading in chunks of 1 million rows at a time, to avoid memory issues
title_principals_pieces = pd.read_csv('../data/raw/imdb/title.principals.tsv.gz',
            sep='\t', na_values='\\N',
            usecols=['tconst', 'nconst', 'category'], chunksize=1_000_000)
# we take the first chunk of 1 million rows and 
# display the first 5 rows and the unique values 
# of the 'category' column
premier = next(title_principals_pieces)
print(premier.head(5))
print(premier['category'].unique())
```

# Mapping of links movies with title.principals movies

```python
# we import the links_df dataframe from the processed data folder and check which tconst values are missing in the title.principals.tsv.gz file
links_df = pd.read_parquet('../data/processed/links_df.parquet')
to_look = set(links_df['tconst'])
found = set()
# we read the title.principals.tsv.gz file in chunks of 1 million rows at a time, and check which tconst values are found in the to_look set
chunks = pd.read_csv('../data/raw/imdb/title.principals.tsv.gz',
                       sep='\t', na_values='\\N',
                       usecols=['tconst'], chunksize=1_000_000)
# we look up in each chunk the tconst values (movie id in links file) 
# that are in the to_look set and add them to the found set
for chunk in chunks:
    found |= to_look & set(chunk['tconst'])
# we easily compare the two sets to_look and found
# which gives us the tconst values of links file
# that are not found in title.principals.tsv.gz file
missing = to_look - found
print(f"{len(found)} found, {len(missing)} missing out of {len(to_look)}")
print(list(missing)[:10])
```
