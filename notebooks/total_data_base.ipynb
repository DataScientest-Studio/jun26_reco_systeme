# Fusion of ratings and movies

```python
import pandas as pd
# we import the processed dataframes
movies_df_mb = pd.read_parquet('../data/processed/movies_df.parquet')
ratings_df = df = pd.read_parquet('../data/processed/ratings_df.parquet')

#we merge the ratings and movies dataframes on 'movieId' 
# to create a combined dataframe
#merge is done with a left join to keep all ratings, 
# even if some movies could be missing in the movies dataframe
ratings_movies_df = ratings_df.merge(movies_df_mb, on='movieId', how='left')

```

```python
# we check the structure of the merged dataframe 
# and the number of missing values in it
print("structure of ratings_df and of ratings_movies_df:", ratings_df.shape, ratings_movies_df.shape)          # même nombre de lignes
print("number of missing values in columns of ratings_movies_df:", ratings_movies_df.isna().sum())    
print("total number of missing values in ratings_movies_df:", ratings_movies_df.isna().sum().sum()) 
display(ratings_movies_df.head(5))
```

# Fusion of genome_scores and genome_tags

```python
# we import the genome scores and genome tags dataframes
genome_scores_df = pd.read_parquet('../data/processed/genome_scores_df.parquet')
genome_tags_df = pd.read_parquet('../data/processed/genome_tags_df.parquet')

#we merge the genome scores and genome tags dataframes on 'tagId'
#with a left join to keep all genome scores, 
# even if some tags could be missing in the genome tags dataframe
genome_merge_df = genome_scores_df.merge(genome_tags_df, on="tagId", how="left")
display(genome_merge_df.head(10))
print("number of missing values:", genome_merge_df.isnull().sum().sum())
```

# Pivot of genome to movies_profile

```python
# we inversed the merged dataframe genome_merge_df 
# to have one line per movie and one column per tag, 
# with the relevance as values
movies_profile_df = genome_merge_df.pivot(index="movieId",columns = "tag",values = "relevance")
print("number of columns in movies_profile_df:", movies_profile_df.shape[1])
# below, we found the same number of movies in movies_profile_df 
# as in genome_scores_df and no missing values.
#  Merge and pivot operations went well
print("number of lines in movies_profile_df:", movies_profile_df.shape[0])
print("number of missing values in movies_profile_df:", movies_profile_df.isnull().sum().sum())
display(movies_profile_df.head(5))
```

```python
#even if merge of ratings_movies_df and movies_profile_df could not be done,
#  we can check if all movies of ratings_movies_df are present in movies_profile_df
to_look = set(ratings_movies_df['movieId'])
found = set()
found = to_look & set(movies_profile_df.index)
missing =to_look-found
print("number of movies of ratings_movies_df missing in movies_profile_df: ", len(missing))
# we have 6426 movies of ratings_movies_df missing in movies_profile_df
```
