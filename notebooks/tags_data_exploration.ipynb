# Import of tags.csv file

```python
import pandas as pd
tags_df = pd.read_csv('../data/raw/tags.csv')
display(tags_df.head())
print("number of missing values in tags_df:",tags_df.isnull().sum().sum())
print("number of unique values of movieId:",tags_df['movieId'].nunique())
print("number of unique values of userId:",tags_df['userId'].nunique())
print("number of lines in tags_df:",tags_df.shape[0])
```

# Data Exploration of tags

```python
import matplotlib.pyplot as plt
nb_tags = tags_df.groupby('userId').size()

plt.hist(nb_tags, bins=50)
plt.yscale('log')          # quelques utilisateurs très actifs écrasent le reste
plt.xlabel('Nombre de tags par utilisateur')
plt.ylabel("Nombre d'utilisateurs")
plt.show()
```
