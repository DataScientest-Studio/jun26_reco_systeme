# Import of genome-tags.csv file

```python
import pandas as pd
genome_tags_df = pd.read_csv('../data/raw/genome-tags.csv')
display(genome_tags_df.head())
print("number of missing values in genome_tags_df:",genome_tags_df.isnull().sum().sum())
print("number of unique values in tagId:",genome_tags_df['tagId'].nunique())
print("number of lines in genome_tags_df:",genome_tags_df.shape[0])
```

# Export of genome_tags_df to parquet file

```python
genome_tags_df.to_parquet('../data/processed/genome_tags_df.parquet')
```
